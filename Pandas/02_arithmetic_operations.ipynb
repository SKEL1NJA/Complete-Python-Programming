```python
import pandas as pd
```

```python
var = pd.DataFrame({"A": [10,20,30,40,50], "B":[60,70,80,90,100]})
print(var)
```

```python
var["C"] = var["A"] + var["B"]
print(var)
```

```python
var["D"] = var["B"] - var["A"]
print(var)
```

```python
var["E"] = var["B"] * var["A"]
print(var)
```

```python
var["F"] = var["B"] / var["A"]
print(var)
```
