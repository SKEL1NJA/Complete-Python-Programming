Series -> one dimensional array that is capable of storing various data types

```python
import pandas as pd
```

```python
var = [12,43,29,8,76]

x = pd.Series(var)
print(x)
print()
print(type(x))
```

```python
var = [12,43,29,8,76]

x = pd.Series(var, index=['a','b','c','d','e'])
print(x)
```

```python
print(x['c'])
```

DataFrame -> Used to store 2D data

```python
var = [12,43,29,8,76]

x = pd.DataFrame(var)
print(x)
print()
print(type(x))
```

```python
var1 = {"a": [1,2,3,4,5], "b": [6,7,8,9,10]}

y = pd.DataFrame(var1)
print(y)
```

```python
var1 = {"a": [1,2,3,4,5], "b": [6,7,8,9,10]}

y = pd.DataFrame(var1, index=['a','b','c','d','e'])
print(y)
```
