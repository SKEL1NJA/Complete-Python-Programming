```python
import pandas as pd
```

```python
var = pd.DataFrame({"A": [1,2,3,4,5], "B":[9,8,7,6,5]})
print(var)
```

### Insert Data

```python
var.insert(1, "Name", [11, 12, 13, 14, 15])

print(var)
```

```python
var["Slicing"] = var["A"][:3]
print(var)
```

### Delete Data

```python
del var["B"]

print(var)
```

```python
var1 = var.pop("Name")

print(var1)
```
