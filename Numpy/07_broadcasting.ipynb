### Rule 2: Start comparing from right side
### Atleast one of the arrays must contain a "1" or both must be same
### The new matrix will be of size of maximum of shapes of both

```python
x = np.array([1,2,3])
y = np.array([[1],[2],[3]])

print(x)
print(y)
print(x.shape)
print(y.shape)

z = x+y
print(z)
print(z.shape)
```
