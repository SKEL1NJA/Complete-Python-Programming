```python
import numpy as np
```

### Shape

```python
var = np.array([[2,3],[2,3]])

print(var)
print(var.shape)
```

```python
var1 = np.array([1,2,3,4], ndmin=5)

print(var1)
print(var1.shape)
```

### Reshape

```python
var2 = np.array([1,2,3,4,5,6])
print(var2.ndim)
print(var2.shape)

print()

x = var2.reshape(3, 2)
print(x)
print(x.ndim)
print(x.shape)
```

```python
one = var2.reshape(-1)
print(one)
print(one.ndim)
print(one.shape)
```
