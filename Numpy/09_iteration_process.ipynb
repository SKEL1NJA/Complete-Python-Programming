```python
import numpy as np
```

### normal looping

```python
arr = np.array([4,5,6,7])

for i in arr:
    print(i)
```

```python
arr1 = np.array([[4,5,6],[7,8,9]])

for i in arr1:
    for j in i:
        print(j)
```

```python
arr2 = np.array([[[3,2,1],[1,2,3],[4,5,6]]])

for i in arr2:
    for j in i:
        for k in j:
            print(k)
```

### using functions

```python
arr3 = np.array([[[7,8,9],
                  [6,5,4],
                  [3,2,1]]])

for val in np.nditer(arr3):
    print(val)
print()
for idx, val in np.ndenumerate(arr3):
    print(idx, val)
```
