```python
import numpy as np
```

### Indexing

```python
oneD = np.array([7,6,9])

print(oneD[1])
print(oneD[-2])
```

```python
twoD = np.array([[7,8,9],
                 [6,5,4],
                 [3,2,1]])

print(twoD[1][1])
```

```python
threeD = np.array([[[8,4,2],
                    [5,3,1]]])

print(threeD[0][0][2])
print(threeD[0][1][1])
```
