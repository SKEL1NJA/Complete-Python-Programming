```python
import numpy as np
```

```python
mat1 = np.matrix([[1, 2, 3], [4, 5, 6]])
mat2 = np.matrix([[10, 11], [12, 13], [14, 15]])

print(mat1)
print(mat1.shape)
print()
print(mat2)
print(mat2.shape)
```

```python
print(mat1*mat2)
print()
print(mat1.dot(mat2))
```

### Matrix Functions

```python
# Transpose

mat = np.matrix([[1,2,3],[4,5,6]])

print(mat)
print()
print(np.transpose(mat))
print()
print(mat.T)
```

```python
# Inverse

a = [[1,2],[3,4]]

print(a)
print()
print(np.linalg.inv(a))
```

```python
# Determinant

b = np.matrix([[1,2,3],[4,5,6],[7,8,9]])

print(b)
print()
print(np.linalg.det(b))
```
